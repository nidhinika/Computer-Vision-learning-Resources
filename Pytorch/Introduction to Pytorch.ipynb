Tensor Reshaping

```python
flat_tensor=torch.rand(784)
```

```python
viewed_tensor=flat_tensor.view(1,28,28)
```

```python
viewed_tensor.shape
```

```python
hwc_tensor=torch.rand(640,480,3)
chw_tensor=hwc_tensor.permute(2,0,1)
chw_tensor.shape
```
